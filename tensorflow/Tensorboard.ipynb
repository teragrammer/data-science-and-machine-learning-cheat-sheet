# What is TensorBoard?

TensorBoard is a visualization toolkit for TensorFlow, used to monitor and analyze machine learning experiments. It helps track metrics like loss and accuracy, visualize model graphs, view histograms of weights/biases, and more.

```python
import tensorflow as tf
from tensorflow.keras.callbacks import TensorBoard
import numpy as np
import datetime
```

```python
tf.__version__
```

```python
# create a simple dataset (for testing purposes only)
x_train = np.array([[0], [1], [2], [3]], dtype=float)
y_train = np.array([[0], [2], [4], [6]], dtype=float) # y = 2x
```

```python
# build a simple model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1])
])

# compile model
model.compile(optimizer="sgd", loss="mean_squared_error")
```

```python
# set up TensorBoard callback
log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorboard_callback = TensorBoard(log_dir=log_dir, histogram_freq=1)
```

```python
# train model
model.fit(x_train, y_train, epochs=50, callbacks=[tensorboard_callback])
```
