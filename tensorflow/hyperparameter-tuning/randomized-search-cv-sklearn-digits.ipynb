We'll use `RandomizedSearchCV` from scikit-learn to tune the hyperparameters of a neural network.

**Steps:**

1. Create a simple neural network model.
2. Define the hyperparameters we want to tune.
3. Use `RandomizedSearchCV` to search for the best hyperparameters.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
```

```python
# load the Iris dataset
data = load_iris()
X = data.data
y = data.target

# split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# define the model
rf_model = RandomForestClassifier(random_state=42)
```

```python
# define the hyperparameter grid
param_dist = {
    'n_estimators': np.arange(50, 200, 50),  # Number of trees in the forest
    'max_depth': np.arange(3, 15, 1),        # Maximum depth of the tree
    'min_samples_split': np.arange(2, 10, 1), # Minimum samples required to split a node
    'min_samples_leaf': np.arange(1, 10, 1),  # Minimum samples required at each leaf node
    'max_features': ['sqrt', 'log2'],  # Number of features to consider for the best split
    'bootstrap': [True, False]  # Whether bootstrap samples are used when building trees
}
```

```python
# set up the RandomizedSearchCV
random_search = RandomizedSearchCV(estimator=rf_model,
                                   param_distributions=param_dist,
                                   n_iter=100,  # Number of random combinations to try
                                   cv=3,        # Cross-validation splitting
                                   verbose=2,   # Print out the progress
                                   random_state=42,
                                   n_jobs=-1)   # Use all available CPU cores
```

```python
# Fit the model to the training data
random_search.fit(X_train, y_train)
```

```python
# Print the best hyperparameters found
print("Best Hyperparameters: ", random_search.best_params_)

# Evaluate the model on the test set
best_model = random_search.best_estimator_
y_pred = best_model.predict(X_test)
test_acc = accuracy_score(y_test, y_pred)
print(f"Test Accuracy: {test_acc}")
```
